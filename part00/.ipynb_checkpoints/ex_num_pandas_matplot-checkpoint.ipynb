```python
#Simulate 1000 days of percentage returns using normal distribution

import numpy as np

np.random.seed(42)
daily_returns = np.random.normal(loc=0.001, scale=0.02, size=1000)

price_path = 100 * np.cumprod(1 + daily_returns)

print(f"Generated an array of {len(price_path)} data points")
```

```python
import pandas as pd

df = pd.DataFrame({'Price': price_path})

df['50_Day_SMA'] = df['Price'].rolling(window=50).mean()

df.tail()
```

```python
#let's visualize the output

import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.plot(df['Price'], label="Asset Price")
plt.plot(df['50_Day_SMA'], label='50 Day SMA', color='red', linewidth=2)

plt.title("Simulated Asset Price with 50 Day Moving Average")
plt.xlabel("Days")
plt.ylabel("Price")
plt.legend()

plt.show()
```
