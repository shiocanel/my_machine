```python
from datasets import load_dataset

print("Downloading dataset...")
dataset = load_dataset("stanfordnlp/imdb", split="train")

split = dataset.train_test_split(test_size=0.2, seed=42)

train_ds = split["train"]
test_ds = split["test"]

print(f"Training exemples: {len(train_ds)}")
print(f"Test exemples: {len(test_ds)}")

print("\nFirst training review excerpt:")
print(train_ds[0]['text'][:200] + "...")
```

```python
# let's see what happens with Parquet

import os

print("Saving to CSV...")
train_ds.to_csv("imdb_train.csv")

print("Saving to Parquet...")
train_ds.to_parquet("imdb_train.parquet")

csv_size = os.path.getsize("imdb_train.csv") / (1024 * 1024)
parquet_size = os.path.getsize("imdb_train.parquet") / (1024 * 1024)

print(f"\nCSV File Size: {csv_size:.2f} MB")
print(f"Parquet File Size: {parquet_size:.2f} MB")
```
